# **Data Cleaning**

## **Create Copy**

```python
df_copy = df.copy()
print('Copy Created boss')
```

## **Data Type**

```python
df_copy['age'] = df_copy['age'].replace("_err",'',regex=True).astype('float')
```

## **Impossible Values**

```python
df_copy.loc[df_copy['age']<18,'age'] = np.nan
df_copy.loc[df_copy['age']>100,'age'] = np.nan

df_copy.loc[df_copy['num_medications']>20,'num_medications'] = np.nan

df_copy.loc[df_copy['glucose_level']>300,'glucose_level'] = np.nan

df_copy.loc[df_copy['bmi']<0,'bmi'] = np.nan

print("Impossible values has been removed boss")
```

## **Duplicates**

```python
print('Duplicates Values: ')
print(f"Before: {df_copy.duplicated().sum()}")
df_copy.drop_duplicates(inplace=True)
print(f"After: {df_copy.duplicated().sum()}")
```

## **NaN values**

### **Fill them**

```python
print("NaN values in the dataset before cleaning is: ")
print(df_copy.isna().sum())
# I am using this method becus it best suited with this dataset 
NaN_columns = df_copy.columns.tolist()
NaN_columns.remove('readmitted_30days')
for col in NaN_columns:
    df_copy[col] = df_copy[col].fillna(floor(df_copy[col].median()))
print("NaN values in the dataset after cleaning is: ")
print(df_copy.isna().sum())
```

### **Drop Them**

```python
print("NaN values in the dataset before Droppig is: ")
print(df_copy.isna().sum())
df_copy.dropna(inplace=True)
print("NaN values in the dataset After Droppig is: ")
print(df_copy.isna().sum())
```
